This project builds an ML OCR model for reading national ID cards and passport

```python
# import modules
import cv2
import pytesseract
from PIL import Image
from matplotlib import pyplot as plt
```

```python
# load image
img_path = "../../images/bene-passeport.png"
img = cv2.imread(img_path)
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB) # convert to RGB for matplotlib
```

```python
# convert image to grayscale and remove color info
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
# print("image in grrayscale")
cv2.imshow("Grayscale Image", gray)
```

```python
# display image - original
plt.figure(figsize=(10, 10))
plt.imshow(img_rgb)
plt.title("Original Image")
plt.axis("off")
plt.show()
```

```python
# display image in grayscale
plt.figure(figsize=(10, 10))
plt.imshow(gray, cmap='gray')
plt.title("Grayscale Image")
plt.axis("off")
plt.show()
```

```python
pytesseract.pytesseract.tesseract_cmd = r"C:/Program Files/Tesseract-OCR/tesseract.exe"
```

```python
# extract text from image using pytesseract
text = pytesseract.image_to_string(img_rgb)
print("Extracted Text:\n")
print(text)
```
